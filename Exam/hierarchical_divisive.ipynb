# Simple Divisive Clustering

Top-down clustering by repeatedly splitting the largest cluster with K-Means.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.metrics import silhouette_score
```

```python
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=1.0, random_state=42)

def divisive_clustering(X, target_k=4):
    clusters = [np.arange(len(X))]

    while len(clusters) < target_k:
        largest_idx = max(range(len(clusters)), key=lambda i: len(clusters[i]))
        points_idx = clusters.pop(largest_idx)

        km = KMeans(n_clusters=2, random_state=42, n_init=10)
        split_labels = km.fit_predict(X[points_idx])

        c0 = points_idx[split_labels == 0]
        c1 = points_idx[split_labels == 1]
        clusters.extend([c0, c1])

    labels = np.zeros(len(X), dtype=int)
    for i, idxs in enumerate(clusters):
        labels[idxs] = i
    return labels

labels = divisive_clustering(X, target_k=4)
score = silhouette_score(X, labels)
print('Silhouette Score:', round(score, 4))
```

```python
plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='tab10', s=30)
plt.title('Divisive Clustering (Top-Down)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```
