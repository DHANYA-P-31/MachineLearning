# Simple HMM (Predefined Functions)

Train an HMM using hmmlearn, test Forward and Viterbi, and save model for `hmmui.py`.

```python
from pathlib import Path
import joblib
import numpy as np
from hmmlearn.hmm import GaussianHMM
```

```python
obs_to_num = {'normal': 0.0, 'cold': 1.0, 'dizzy': 2.0}

sequences = [
    ['normal', 'normal', 'cold', 'dizzy'],
    ['normal', 'cold', 'cold', 'dizzy'],
    ['normal', 'normal', 'normal', 'cold'],
    ['cold', 'dizzy', 'dizzy', 'dizzy']
]

X_list = []
lengths = []
for seq in sequences:
    arr = np.array([obs_to_num[s] for s in seq], dtype=float).reshape(-1, 1)
    X_list.append(arr)
    lengths.append(len(arr))

X_train = np.vstack(X_list)

model = GaussianHMM(n_components=2, covariance_type='diag', n_iter=300, random_state=42)
model.fit(X_train, lengths)

state_names = [f'State {i}' for i in range(model.n_components)]
print('Model trained. States:', state_names)
```

```python
test_seq = ['normal', 'cold', 'dizzy']
X_test = np.array([obs_to_num[s] for s in test_seq], dtype=float).reshape(-1, 1)

# Forward probability from built-in score() (returns log probability)
log_prob = model.score(X_test)
forward_prob = float(np.exp(log_prob))

# Viterbi path from built-in decode()
v_log_prob, v_states = model.decode(X_test, algorithm='viterbi')
v_path = [state_names[s] for s in v_states]

print('Test sequence:', test_seq)
print('Forward probability:', forward_prob)
print('Viterbi path:', v_path)
print('Viterbi log probability:', float(v_log_prob))
```

```python
artifact = {
    'model': model,
    'obs_to_num': obs_to_num,
    'state_names': state_names
}

model_path = Path('hmm_model.joblib')
joblib.dump(artifact, model_path)
print('Saved:', model_path.resolve())
```
