# SVM Spam Detection (Minimal)
Simple pipeline: load data, train TF-IDF + LinearSVC, evaluate, and test sample messages.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, classification_report
import joblib
```

```python
df = pd.read_csv(r"../data/spamhamdata.csv", sep="\t", header=None, names=["label", "text"])
df = df.dropna()
print("Dataset shape:", df.shape)
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df["text"], df["label"], test_size=0.2, random_state=42, stratify=df["label"]
)

vectorizer = TfidfVectorizer(stop_words="english")
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

model = LinearSVC()
model.fit(X_train_vec, y_train)
print("Model trained")
```

```python
y_pred = model.predict(X_test_vec)
acc = accuracy_score(y_test, y_pred)
print(f"Accuracy: {acc:.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
sample_texts = [
    "Congratulations! You have won a free ticket. Reply WIN now.",
    "Hey, are we still meeting for lunch today?"
]
sample_vec = vectorizer.transform(sample_texts)
sample_pred = model.predict(sample_vec)
for text, label in zip(sample_texts, sample_pred):
    print(f"{label.upper()}: {text}")
```

```python
artifact = {
    "model": model,
    "vectorizer": vectorizer
}
joblib.dump(artifact, "svm_model.joblib")
print("Saved model artifact to svm_model.joblib")
```
